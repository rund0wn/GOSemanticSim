### Semantic Similarity

`Annotation set`: all GOA UniProt proteins, not including electronically inferred annotations by filtering out evidence code `IEA`

This script calculates the following:
- `Information Content (IC)`: how informative is the GO term *(based on annotation set and DAG)*
- `Semantic Similarity`: distance between two terms based on the DAG *(NOT based on annotation set and DAG)*
- `Resnik's Similarity`: IC of the most informative common ancestor; most specific common parent in DAG *(based on annotation set and DAG)*
- `Lin's Similarity`: similar to Resnik's, but divided by the sum of information of the two individual terms *(based on annotation set and DAG)*

*Use Lin's Similarity to compare across sets*

#### Import packages

```python
import os
from goatools.obo_parser import GODag
from goatools.anno.gaf_reader import GafReader
import goatools.semantic as sem
```

#### Load input files
- GAF annotations from GOA
- GO DAG

**NOTE: this is specific to the sub-ontology, change namespace to calculate for a sub-ontology other than `MF`.**

```python
## Load files
uniprot_gaf = ("../go_data/filtered_goa_uniprot_all_noiea.gaf") # Annotation set
godag = GODag("../go_data/go.obo") # GO graph (DAG)

## Create GOATools objects
ogaf = GafReader(uniprot_gaf) # Read annotation file
gene2gos = ogaf.get_id2gos(namespace='MF') # Create gene to GO term mapping for Molecular Function (MF) namespace
```

#### Term counts/frequencies
Create term count object based on annotation set and GO DAG

```python
## Term counts for calculations
termcounts = sem.TermCounts(godag, gene2gos) # Calculate term counts for the GO graph and gene annotations

## Example MF GO terms for testing
go_term1 = 'GO:0005488' # MF term for "binding"
go_term2 = 'GO:0044877' # MF term for "protein-containing complex binding"

## Test term counts and frequencies
count1 = termcounts.get_count(go_term1)
count2 = termcounts.get_count(go_term2)
freq1 = termcounts.get_term_freq(go_term1)
freq2 = termcounts.get_term_freq(go_term2)

print(f"Count: {go_term1} = {count1}, {go_term2} = {count2}")
print(f"Frequency: {go_term1} = {freq1}, {go_term2} = {freq2}")
```

#### Calculate Information Content

```python
## Information content calculations for the example MF GO terms
infocontent = sem.get_info_content(go_term1, termcounts)
print(f"Information content of {go_term1}: {infocontent}")

infocontent2 = sem.get_info_content(go_term2, termcounts)
print(f"Information content of {go_term2}: {infocontent2}")
```

#### Calculate Semantic Similarity

```python
## Semantic similarity calculations for the example MF GO terms

sim = sem.semantic_similarity(go_term1, go_term2, godag) # Semantic Similarity based on the GO graph structure
print(f"Semantic similarity: {sim}")

sim_r = sem.resnik_sim(go_term1, go_term2, godag, termcounts) # Resnik's Similarity
print(f"Resnik similarity: {sim_r}")

sim_l = sem.lin_sim(go_term1, go_term2, godag, termcounts) # Lin's Similarity
print(f"Lin similarity: {sim_l}")
```
